Do not replace the preprocessor with manual normalization (such as dividing by 255), because it may produce incorrect predictions.

```python
# Import Libraries
import tensorflow as tf
from tensorflow import keras
from keras import layers
import keras_hub #provides pretrained models like: ViT, Bert
import numpy as np
from PIL import Image

# Load ViTImageClassifierPreprocessor (vit_base_patch16_224_imagenet preset)

preprocessor = keras_hub.models.ViTImageClassifierPreprocessor.from_preset(
    "vit_base_patch16_224_imagenet"
)

# Load ViTImageClassifier (vit_base_patch16_224_imagenet preset)

classifier = keras_hub.models.ViTImageClassifier.from_preset(
    "vit_base_patch16_224_imagenet",
    num_classes=1000,
    preprocessor=preprocessor,
)
# Load the images

image_paths = [
    "/Users/layan/Documents/GitHub/arti560-computer-vision-labs/lab03-image-classification-vision-transformer-vit/bow.jpg",
    "/Users/layan/Documents/GitHub/arti560-computer-vision-labs/lab03-image-classification-vision-transformer-vit/jellyfish.jpg",
    "/Users/layan/Documents/GitHub/arti560-computer-vision-labs/lab03-image-classification-vision-transformer-vit/Kite.jpg",
    "/Users/layan/Documents/GitHub/arti560-computer-vision-labs/lab03-image-classification-vision-transformer-vit/Valley.jpg",
    "/Users/layan/Documents/GitHub/arti560-computer-vision-labs/lab03-image-classification-vision-transformer-vit/purse.jpeg",
]

true_labels = [
    "bow",
    "jellyfish",
    "Kite",
    "Valley",
    "purse",
]

images = []
for path in image_paths:
    img = Image.open(path).convert("RGB")
    images.append(np.array(img))

# Load ImageNet class names
imagenet_classes = open("/Users/layan/Documents/GitHub/arti560-computer-vision-labs/lab03-image-classification-vision-transformer-vit/imagenet_classes.txt").read().splitlines()

# Predict classes
for path, img_array, true_label in zip(image_paths, images, true_labels):
    preds = classifier.predict(np.expand_dims(img_array, axis=0))
    pred_idx = np.argmax(preds, axis=-1)[0]
    predicted_label = imagenet_classes[pred_idx]
    print(f"File: {path.split('/')[-1]:<15}  Predicted: {predicted_label:<25}  True: {true_label}")
```
